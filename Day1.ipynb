A var is a container that can hold any type of data:

var = 10 ---> Literal

number DT - stores multiple value , immutable , means changing the values DT result in newly allocated object

1. int (Integer)
   - Stores whole numbers (positive, negative, or zero).
   - Examples: 10, -25, 0
   - Example:
       a = 100

2. float (Floating Point)
   - Stores numbers with decimal points.
   - Examples: 3.14, -2.5, 0.0
   - Example:
       b = 3.14

3. complex (Complex Number)
   - Stores numbers in the form a + bj, where:
       a = Real part
       b = Imaginary part
       j = square root of Imaginary unit
   - Examples: 2+3j, 5j
   - Example:
       c = 2 + 3j

Type Conversion:
   int()      -> Converts a value to an integer.
   float()    -> Converts a value to a floating-point number.
   complex()  -> Converts a value to a complex number.

Examples:
   int(3.9)      # 3
   float(10)     # 10.0
   complex(5)    # (5+0j)
